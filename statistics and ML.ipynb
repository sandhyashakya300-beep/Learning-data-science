```python
# confusion Matrix
#             || predicted +ve || predicted -ve ||
# Actual +ve ||     TP         ||     FN        ||
#Actual -ve   ||    FP         ||     TN        ||

#TP(True Positive): Model correctly predicts positives               {spam correctly detected}
#TN(True Negative): Model correctly predicts negatives                {normal email correctly identified}
#FP(False Positive): Model predicts positive but actually negative    {important email marked as spam}
#FN(False Negative): Model predicts negative but actually positive     {spam missed}

```

```python
# Accuracy 

#Accuracy = (TP + TN) / (TP + TN + FP + FN)

# It measure overall correctness
# If data is imbalanced
# Total 100 emails : 95 normal and 5 spam
#Model predicts 100 normal
#Accuracy = 
#TP:95
#TN:0
#FP:5
#FN:0

acc = (95+0)/(95+0+5+0)
print(acc)
```

```python
#1.precision = TP / (TP + FP)
precision = 95/(95+5)
print(precision)
```

```python
# Recall = TP / (TP+FN)
recall = 95/(95+0)
print(recall)
```

```python
#F1 Score = 2 * (precision * recall) / (precision + recall)
f1_score = 2 * (precision * recall) / (precision + recall)
print(f1_score)
```

```python
# Type of unsupervised learning 
# clustering learning : grouping of similar data

# Association Rule Learning : finding relationships between variables 
# Dimensionality Reduction : reducing number of features 

# K-Means clustering
#It try to divide a dataset into k group 
# Each group has a center called centeriod.

#K-Means minimizes : wcss(Within cluster sum of squares)
# GOal : make the cluster tight and compact
```

```python
# Code implementation 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
from sklearn.datasets import make_blobs

X , y = make_blobs(n_samples=500,centers=6,cluster_std=0.6,random_state=42)
```

```python
plt.scatter(X[:,0],X[:,1])
plt.title("raw data")
plt.show()
```

```python
# Applying K-Mean

kmeans = KMeans(n_clusters=3,random_state=42)
kmeans.fit(X)
```

```python
labels = kmeans.labels_

centeroid = kmeans.cluster_centers_

```

```python
plt.scatter(X[:,0],X[:,1],c=labels)
plt.scatter(centeroid[:,0], centeroid[:,1], color="blue", marker="x", s=200)
plt.title("kMeans Clustering")
plt.show()
```

```python
# elbow mehtod
wcss = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i,random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,11),wcss)
plt.title("elbow method")
plt.xlabel(" no of clusters")
plt.ylabel("wcss")
plt.show()
```

```python
#PCA (principle components analysis)
# /PCA creates new features we call them principle components.
# Date : Year, Quarter, MOnth , Week , Day
# quantity  || Unit_Price  ||  amount
# 5         ||   50        ||  250

# sum(amount) = revenue
```

```python
# Mean centering in the data 
# X_centered = X - mu(mean)

#Covariance matrix
#positive covariance: both of the features increases
# negative covariance: one is decreasing and other is increasing
# zero covariance: there is no relationships between variables

# sigma = (1/(n-1))*(X_centered^T X_centered)

#Eigenvalues and Eigenvector
# v ( Eigenvector) a direction in feature space
# lamda( Eigenvalues) : amount of variance captured in that direction 

# if lamda is large: it capture the high variance.
# IF lamda is small : it captures a low variance.
# we sort Eigenvector in descending of  Eigenvalues.
```

```python
# code implmentation 
import numpy as np 

X = np.array([
    [2.5,2.4],
    [0.5,0.7],
    [2.2,2.9],
    [1.9,2.2],
    [3.1,3.0],
    [2.3,2.7],
    [2.0,1.6],
    [1.0,1.1],
    [1.5,1.6],
    [1.1,0.9]
    ])

# print the dimension
print(X.ndim)
```

```python
# step 1: mean centering
X_mean = np.mean(X,axis=0)
X_centered = X - X_mean
print(X_centered)
```

```python
# Step 2 : covariance matrix 

cov_matrix = np.cov(X_centered.T)
print(cov_matrix)
```

```python
# Step 3 : Eigen decomposition

eigenvalues,eigenvectors = np.linalg.eig(cov_matrix )
print(eigenvalues)
print(eigenvectors)
```

```python
# Step 4: sort eigenvalues

sorted_index = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[sorted_index]
sorted_eigenvectors = eigenvectors[:,sorted_index]

print(sorted_eigenvalues)
print(sorted_eigenvectors)
```

```python
# Step 5 : Top k 

k = 1
w = sorted_eigenvectors[:,:k]
```

```python
# Step 6: projection 
X_reduced = np.dot(X_centered,w)
print(X_reduced)

# printing the outputs
print("Original shape:",X.shape)
print("Reduced shape:",X_reduced.shape)
```

```python
# BOW ( Bag of words)

from sklearn.feature_extraction.text import CountVectorizer
doc = [
    'I Love NLP',
    "I LIke Python",
    "I Love Spicy Food",
    'I LIke Horror MOVIES'
]
vect = CountVectorizer()
X = vect.fit_transform(doc)
print("vocab:",vect.get_feature_names_out())
print("BOW Matrix :\n",X.toarray())
```

```python
# TF-IDF (Term Frequency - Inverse Document Frequency)

from sklearn.feature_extraction.text import TfidfVectorizer
doc = [
    'I Love NLP',
    'I LIke Python',
    'I Love Spicy Food',
    'I LIke Horror MOVIES'
]
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(doc)
print("vocab:",tfidf.get_feature_names_out())
print("tfidf Matrix :\n",X.toarray())

```
